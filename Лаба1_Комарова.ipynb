### **Условие задачи**
Сноубордист, движущийся в горизонтальном направлении со скоростью $V_0$, начинает подниматься по дуге окружности радиуса R = 9 м. Определить минимальное значение $V_0$, при котором сноубордист поднимется на вертикальную стенку. Коэффициент трения скольжения сноуборда по дуге μ = 0,1.

### **1. Содержательная постановка задачи.**

Требуется разработать математическую модель, позволяющую описать движение сноубордиста по дуге окружности до вертикальной оси с учетом сил тяжести и трения.

**Модель должна позволять:**
*   Определять скорость сноубордиста в любой точке траектории на дуге.
*   Найти зависимость между начальной скоростью сноубордиста и высотой его подъема.
*   Определить минимальную начальную скорость, достаточную для достижения верхней точки дуги (вертикальной стенки).

**Исходные данные:**
*   Форма траектории подъема – дуга окружности радиусом R.
*   Начальная скорость $V_0$, направленная горизонтально.
*   Коэффициент трения скольжения μ.
*   Конечная точка траектории – вертикальная стенка, что соответствует подъему на высоту, равную радиусу R.

### **2. Концептуальная постановка задачи.**

Движение сноубордиста по дуге окружности может быть описано в рамках законов динамики Ньютона или с использованием закона сохранения энергии с учетом работы диссипативных сил (силы трения).

**Применим следующие гипотезы:**

*   Объектом моделирования является сноубордист, которого мы рассматриваем как материальную точку.
*   Движение происходит в инерциальной системе отсчета, связанной с Землей.
*   Силы, действующие на сноубордиста: сила тяжести mg, сила нормальной реакции опоры N и сила трения скольжения $F_{тр}$. Сопротивлением воздуха пренебрегаем.
*   Сила трения подчиняется закону Амонтона-Кулона: $F_{тр}$ = μN, где μ – постоянный коэффициент трения скольжения, а N – сила нормальной реакции опоры. Сила трения всегда направлена против вектора скорости.
*   Форму траектории идеализируем, представив ее как точную дугу окружности радиуса R. Переход с горизонтального участка на дугу считаем мгновенным и без потерь энергии на удар.
*   Движение происходит в однородном поле силы тяжести с постоянным ускорением свободного падения g (примем g ≈ 9.8 м/с²).


### Обоснование используемых гипотез


**1. Объект моделирования — материальная точка.**

**Обоснование:** Это допущение правомерно, когда размеры объекта значительно меньше характерных размеров его траектории. В задаче радиус дуги **R = 9 м**, что в несколько раз превышает рост сноубордиста (возьмём средний рост ≈1.7-1.9 м). В этом масштабе его собственное вращение или изменение позы (например, взмах руками) вносит незначительный вклад в общую кинетическую энергию по сравнению с энергией поступательного движения его центра масс. Моделирование сноубордиста как твёрдого тела потребовало бы учёта момента инерции и ввело бы излишнюю сложность, не влияющую на принципиальный результат — нахождение минимальной начальной скорости.

**2. Инерциальная система отсчета, связанная с Землей.**

**Обоснование:** Для процессов, происходящих на поверхности Земли в течение короткого времени и на небольших расстояниях, эффекты от вращения планеты пренебрежимо малы. Ускорения, вызываемые этими эффектами, на много порядков меньше ускорения свободного падения g. Поэтому Земля с высокой точностью может считаться инерциальной системой отсчета.

**3. Пренебрежение сопротивлением воздуха.**

**Обоснование:** Задача состоит в поиске *минимальной* скорости. Это означает, что сноубордист достигает верхней точки с нулевой скоростью, то есть на протяжении всего подъёма он замедляется. Скорости, особенно на конечном участке, невелики. Хотя сопротивление воздуха, безусловно, присутствует, его вклад в общие потери энергии считается значительно меньшим, чем работа силы трения о снег и работа против силы тяжести.

**4. Сила трения подчиняется закону Амонтона-Кулона (Fтр = μN).**

**Обоснование:** Это стандартная модель для сухого трения скольжения. Для пары "поверхность сноуборда — снег/лёд" эта модель работает с достаточной точностью.

**5. Идеализированная траектория.**

**Обоснование:** Это позволяет описать положение сноубордиста всего одной переменной — углом φ, что значительно упрощает математическую постановку. Допущение о безударном переходе позволяет сосредоточиться на динамике движения именно по дуге, что и является предметом задачи.

**6. Однородное поле силы тяжести.**

**Обоснование:** Это допущение оправданно, так как максимальный перепад высот в задаче составляет 9 метров. На таком масштабе изменение g из-за изменения расстояния до центра Земли составляет ничтожно малую величину. Гравитационное поле Земли на такой малой площади можно считать полностью однородным.

В соответствии с изложенными гипотезами, состояние системы в любой момент времени полностью определяется положением сноубордиста на дуге и его скоростью. Положение удобно характеризовать углом φ, отсчитываемым от горизонтального направления.

Применим второй закон Ньютона в проекциях на касательное и нормальное направления к траектории или закон изменения кинетической энергии: изменение кинетической энергии тела равно суммарной работе всех действующих на него сил.

d($E_k$) = d($A_g$) + d($A_N$) + d($A_{тр}$)

*   Изменение кинетической энергии на малом участке пути dS: d($E_k$) = d(½mv²).
*   Работа силы тяжести: $A_g$. Компонента силы тяжести, совершающая работу, – это касательная составляющая mg×sin(φ). Работа на пути dS = R×dφ будет отрицательной: d$A_g$ = -mg×sin(φ)dS = -mgR×sin(φ)dφ.
*   Работа силы нормальной реакции опоры A_N равна нулю, так как эта сила всегда перпендикулярна вектору перемещения.
*   Работа силы трения $A_{тр}$ отрицательна: d$A_{тр}$ = -$F_{тр}$×dS = -μN×R×dφ.

Для нахождения силы нормальной реакции N запишем второй закон Ньютона в проекции на нормаль (радиальное направление). Центростремительное ускорение равно $a_ц$ = v²/R. На этом направлении действуют сила нормальной реакции и проекция силы тяжести mg×cos(φ):
N - mg×cos(φ) = mv²/R
Отсюда N = mg×cos(φ) + mv²/R.

Таким образом, концептуальная постановка сводится к следующей задаче:

Определить закон изменения скорости сноубордиста v в зависимости от его углового положения φ, описываемый дифференциальным уравнением, следующим из условия равенства изменения кинетической энергии и работы сил:

d(½mv²) = -mgR*sin(φ)*dφ - μ(mg*cos(φ) + mv²/R)*R*dφ

с начальным условием V(φ=0) = $V_0$. Найти минимальное значение $V_0$, при котором сноубордист достигнет угла φ = π/2 (вертикальная стенка) со скоростью V(φ=π/2) ≥ 0. Минимальное значение $V_0$ будет соответствовать случаю, когда V(φ=π/2) = 0.

### **3. Математическая постановка задачи.**

Законченная концептуальная постановка позволяет сформулировать математическую задачу.

Требуется найти минимальное значение параметра $V_0$, при котором решение дифференциального уравнения, описывающего зависимость скорости v от угла φ, существует на всем интервале φ от 0 до π/2.

Дифференциальное уравнение:

d(½mv²) = -mgR*sin(φ)*dφ - μ(mg*cos(φ) + mv²/R)*R*dφ

Разделив уравнение из концептуальной постановки d(½mv²) = -mgR×sin(φ)×dφ - μ(mg×cos(φ) + mv²/R)×R×dφ на dφ и m, и раскрыв производную, получим:

v(dv/dφ) = -gR×sin(φ) - μ(g×cos(φ) + v²/R)×R

v(dv/dφ) = -gR(sin(φ) + μ×cos(φ)) - μv²

Перегруппируем члены:

v(dv/dφ) + μv² = -gR(sin(φ) + μ×cos(φ))

Это дифференциальное уравнение первого порядка относительно функции v(φ). Удобно сделать замену y = v², тогда dy/dφ = 2v(dv/dφ). Уравнение примет вид линейного неоднородного дифференциального уравнения:

(1/2)(dy/dφ) + μy = -gR(sin(φ) + μ×cos(φ))

или

dy/dφ + 2μy = -2gR(sin(φ) + μ×cos(φ))

**Начальное условие:**
В начальный момент времени, когда сноубордист въезжает на дугу, угол φ = 0, а его скорость равна $V_0$.
v(φ=0) = $V_0$  или  y(φ=0) = $V_0$²

**Краевое условие для нахождения минимальной $V_0$:**
Сноубордист должен достичь вертикальной стенки, что соответствует углу φ = π/2. Минимальная начальная скорость соответствует случаю, когда в этой точке скорость обращается в ноль.
v(φ=π/2) = 0  или  y(φ=π/2) = 0

**Итоговая математическая задача:**

Найти значение $V_0$, при котором решение задачи Коши для дифференциального уравнения dy/dφ + 2μy = -2gR(sin(φ) + μ×cos(φ)) с начальным условием y(0) = $V_0$² удовлетворяет условию y(π/2) = 0.

**Параметры:**
*   R = 9 м
*   μ = 0,1
*   g ≈ 9,8 м/с²

---

###Решение математической задачи (задачи Коши)

`dy/dφ + 2μy = -2gR(sin(φ) + μ*cos(φ))`

Дано ЛНДУ 1 порядка с постоянными коэффициентами и стандартной правой частью. Решим методом суперпозиции.

Будем искать общее решение в виде:

y(φ) = $y_о$(φ) + $y_ч$(φ)

Решим соответствующее однородное уравнение: dy/dφ + 2μy = 0. Это ДУ с разделяющимися переменными. Разделив и проинтегрировав, получим:

$y_о$(φ) = Ce^(-2μφ), где C - произвольная постоянная

Далее найдём частное решение с помощью метода неопределённых коэффициентов. Будем искать его в виде:

$y_ч$ = Asin(φ) + Bcos(φ), тогда d$y_ч$/dφ = Acos(φ) - Bsin(φ)

Подставляя в исходное неоднородное уравнение, получим:

(Acos(φ) - Bsin(φ)) + 2μ(Asin(φ) + Bcos(φ)) = -2gRsin(φ) - 2gRμcos(φ)

Сгруппируем слагаемые при синусе и косинусе:

sin(φ) × (2μA - B) + cos(φ) × (A + 2μB) = sin(φ) × (-2gR) + cos(φ) × (-2gRμ)

Приравняем коэффициенты при синусе и косинусе и получим систему, а затем и сами неизвестные коэффициенты A и B:

2μA - B = -2gR, A + 2μB = -2gRμ

A = -6gRμ / (1 + 4μ²), B = 2gR(1 - 2μ²) / (1 + 4μ²)

Итак, мы нашли частное решение:

$y_ч$(φ) = - (6gRμ / (1 + 4μ²))sin(φ) + (2gR(1 - 2μ²) / (1 + 4μ²))cos(φ)

Таким образом, общее решение исходного уравнения принимает вид:

y(φ) = Ce^(-2μφ) - (2gR / (1 + 4μ²)) × [3μsin(φ) - (1 - 2μ²)cos(φ)]

Применяя граничные условия y(0) = $V_0$² и y(π/2) = 0, получим:

V₀² = (gR/(4μ² + 1)) * (2 - 4μ² + 6μe^(μπ))

Подставляя входные параметры:
*   R = 9 м
*   μ = 0,1
*   g ≈ 9,8 м/с²

получим **минимальное значение начальной скорости**:

**V₀ ≈ 15.359 м/с**

### Проверка по критериям:

#### 1. Контроль размерностей

Проверяемое уравнение: dy/dφ + 2μy = -2gR(sin(φ) + μcos(φ))

*   **y = v²**: Размерность скорости в квадрате `[м²/с²]`.
*   **φ**: Угол, безразмерная величина `[1]`.
*   **dy/dφ**: Имеет ту же размерность, что и y, то есть `[м²/с²]`.
*   **μ**: Коэффициент трения, безразмерный `[1]`.
*   **2μy**: Произведение безразмерной величины на y, итоговая размерность `[L²/T²]`.
*   **g**: Ускорение свободного падения `[м/с²]`.
*   **R**: Радиус `[м]`.
*   **gR**: Произведение, имеющее размерность `[м²/с²]`.
*   **sin(φ), cos(φ)**: Тригонометрические функции, безразмерные `[1]`.

Все слагаемые в уравнении имеют одинаковую размерность `[м²/с²]`.

**Контроль размерностей пройден успешно.**

#### 2. Контроль порядков

*   **Входные параметры:** Радиус R = 9 м — это адекватный размер для рампы. Коэффициент трения μ = 0.1 — разумное значение для пары "сноуборд-снег".
*   **Выходной результат:** Начальная скорость V₀ ≈ 15.359 м/с.
*   **Анализ результата:** Переведем в другую единицу: 15.36 м/с * 3.6 ≈ 55.3 км/ч. Для того чтобы заехать на вертикальную стену высотой 9 метров, преодолевая трение, требуется скорость свыше 50 км/ч на входе, что выглядит правдоподобно, особенно для экстремального спорта.

**Контроль порядков пройден успешно.**

#### 3. Контроль характера зависимостей
Проверяем, как меняется результат V₀ при изменении входных параметров R, g, μ, и соответствует ли это зависимость физическому смыслу.

**Зависимость от R и g:** Из итогового уравнения очевидно, что V₀ растет с увеличением радиуса R и ускорения g (прямопропорциональная зависимость). Это логично, т.к. чтобы заехать выше или бороться с большей силой тяжести, нужна большая начальная скорость.

***Зависимость от μ:** При увеличении трения потеря энергии возрастает. Следовательно, потребуется большая начальная скорость V₀. V₀ прямопропоционально росту коэффициента трения.

**Контроль характера зависимостей пройден успешно.**

#### 4. Контроль экстремальных ситуаций

**Случай 1: Трение отсутствует (μ = 0).**
В этом случае модель должна описывать гладкое скольжение без трения..
Подставляем μ = 0 в итоговую формулу:

V₀² = (gR/(4×0² + 1)) × (2 - 4×0² + 6×0×e^0) = (gR/1) × (2) = 2gR

Получим V₀ = sqrt(2gR). Данная формула соответствует закону сохранения энергии: начальная кинетическая энергия ½mV₀² полностью переходит в потенциальную mgR, работа диссипативных сил в данном случае равна нулю.

**Случай 2: Гравитация отсутствует (g = 0).**

Если g = 0, то V₀² = 0. Если нет силы, которая тянет сноубордиста вниз, то для достижения любой точки на рампе ему не нужна начальная скорость.

**Контроль экстремальных ситуаций пройден успешно.**

#### 5. Контроль граничных условий

**Начальное условие:** v(φ=0) = V₀. Задает начальную скорость в нижней точке дуги.

**Краевое условие:** v(φ=π/2) = 0. Означает, что сноубордист достигает верхней точки φ=π/2 с нулевой скоростью. Так как нам требуется найти минимальную начальную скорость, то нам достаточно, чтобы сноубордист остановился на конечной точке рампы.

**Граничные условия поставлены верно и однозначно определяют искомую величину. Контроль пройден.**

#### 6. Контроль физического смысла

**Силы:** Сила тяжести -mgsin(φ) и сила трения -μN обе направлены против движения (вдоль касательной вниз).

**Сила нормальной реакции N:** N = mgcos(φ) + mv²/R. Она складывается из проекции силы тяжести и центробежной силы. Это выражение следует из второго закона Ньютона в проекции.

**Все компоненты модели основаны на фундаментальных законах физики и корректно их отражают. Контроль пройден.**

#### 7. Контроль математической замкнутости

**Неизвестные:** В задаче две неизвестные — это константа интегрирования C и искомый параметр V₀.

**Уравнения и условия:** У нас есть одно дифференциальное уравнение и два условия.

**Решение:** Общее решение ДУ содержит одну произвольную постоянную C. Используя два граничных условия, мы формируем систему из двух алгебраических уравнений относительно C и V₀. Такая система однозначно определяет обе эти величины.

**Система уравнений и условий является замкнутой и имеет единственное решение для V₀. Контроль математической замкнутости пройден.**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq
from scipy.interpolate import interp1d

# 1. Определение математической модели (правая часть ОДУ)
def ode(phi, y, R, mu, g):
    v_squared = y[0]

    if v_squared < 0:
        v_squared = 0

    dydphi = -2 * mu * v_squared - 2 * g * R * (np.sin(phi) + mu * np.cos(phi))
    return [dydphi]

# 2. Реализация простого численного метода (Метод Эйлера)
class EulerResult:
    def __init__(self, t, y, status):
        self.t = np.array(t)
        self.y = np.array(y)
        self.status = status
        self._interpolator = interp1d(self.t, self.y[0], kind='linear', bounds_error=False, fill_value="extrapolate")

    def sol(self, t_query):
        res = self._interpolator(t_query)
        if np.isscalar(t_query):
            return [res]
        return [res]

def solve_for_v0(v0_initial, R, mu, g):
    y_curr = v0_initial**2
    phi_curr = 0.0
    phi_end = np.pi / 2

    step_size = 0.001

    t_history = [phi_curr]
    y_history = [y_curr]

    status = 0

    while phi_curr < phi_end:
        deriv = ode(phi_curr, [y_curr], R, mu, g)[0]

        y_next = y_curr + deriv * step_size
        phi_next = phi_curr + step_size

        if y_next <= 0:
            y_next = 0
            status = 1
            t_history.append(phi_next)
            y_history.append(y_next)
            break

        if phi_next > phi_end:
            actual_step = phi_end - phi_curr
            y_next = y_curr + deriv * actual_step
            phi_next = phi_end

        y_curr = y_next
        phi_curr = phi_next

        t_history.append(phi_curr)
        y_history.append(y_curr)

    return EulerResult(t_history, [y_history], status)

# 3. Функции для анализа и получения данных от пользователя
def get_user_input():
    try:
        r_in = input("Введите радиус дуги R (метры) [по умолчанию 9.0]: ")
        R = float(r_in) if r_in else 9.0

        mu_in = input("Введите коэффициент трения mu [по умолчанию 0.1): ")
        mu = float(mu_in) if mu_in else 0.1

        g_in = input("Введите ускорение свободного падения g (м/с2) [по умолчанию 9.81): ")
        g = float(g_in) if g_in else 9.81

        if R <= 0 or mu < 0 or g <= 0:
            raise ValueError("Параметры должны быть положительными.")
        return R, mu, g
    except ValueError as e:
        print(f"Ошибка ввода: {e}. Используются значения по умолчанию.")
        return 9.0, 0.1, 9.81

# 4. Основной блок выполнения
if __name__ == "__main__":
    R, mu, g = get_user_input()
    print("\n--- Параметры модели ---")
    print(f"Радиус (R): {R} м")
    print(f"Коэффициент трения (mu): {mu}")
    print(f"Ускорение свободного падения (g): {g} м/с^2\n")

    def final_velocity_sq_objective(v0):
        sol = solve_for_v0(v0, R, mu, g)
        if sol.status == 1:
            return -1
        else:
            final_v_sq = sol.sol(np.pi/2)[0]
            return final_v_sq

    try:
        v0_min = brentq(final_velocity_sq_objective, a=0.1, b=50, xtol=1e-6)
        print("--- Результат ---")
        print(f"Минимальная начальная скорость V0 для достижения стенки: {v0_min:.4f} м/с")
    except ValueError:
        print("Не удалось найти минимальную скорость.")
        exit()

    # График скорости от угла
    sol_min_v0 = solve_for_v0(v0_min, R, mu, g)
    phi_vals = np.linspace(0, np.pi/2, 200)
    y_vals = sol_min_v0.sol(phi_vals)[0]
    v_vals = np.sqrt(np.maximum(0, y_vals))

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)

    plt.plot(np.degrees(phi_vals), v_vals, lw=2)
    plt.title(f'Скорость сноубордиста при $V_0$ = {v0_min:.2f} м/с (Эйлер)')
    plt.xlabel(r'Угол $\phi$, градусы')
    plt.ylabel('Скорость v, м/с')
    plt.grid(True)
    plt.xlim(0, 90)
    plt.ylim(bottom=0)

    # График зависимости высоты подъема
    v0_range = np.linspace(0.1, v0_min * 1.5, 100)
    max_heights = []

    for v0_test in v0_range:
        sol = solve_for_v0(v0_test, R, mu, g)
        max_phi_reached = sol.t[-1]
        max_height = R * np.sin(max_phi_reached)
        max_heights.append(max_height)

    plt.subplot(1, 2, 2)
    plt.plot(v0_range, max_heights, lw=2, label='Максимальная высота подъема')
    plt.axhline(y=R, color='r', linestyle='--', label=f'Целевая высота ({R} м)')

    plt.axvline(x=v0_min, color='g', linestyle='--', label=f'Минимальная $V_0$ = {v0_min:.2f} м/с')

    plt.title('Зависимость высоты подъема от начальной скорости')

    plt.xlabel(r'Начальная скорость $V_0$, м/с')
    plt.ylabel('Максимальная высота подъема h, м')
    plt.grid(True)
    plt.legend()
    plt.ylim(bottom=0)

    plt.tight_layout()
    plt.show()
```
